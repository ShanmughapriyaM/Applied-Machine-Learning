```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshowf
from matplotlib import pyplot as plta

# 1. Load the image
# You can upload an image to your Colab notebook's file system or use a URL.
# For example, let's use an image from a URL.

img = cv2.imread('/content/drive/MyDrive/Applied ML/img1.jpg', 0)  # Read as grayscale

if img is None:
    print("Error: Could not read the image.")
else:
    # 2. Apply Sobel operators
    # dx: Sobel in the X direction
    sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=5)
    # dy: Sobel in the Y direction
    sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=5)

    # 3. Combine the results
    # The magnitude of the gradient is calculated as sqrt(sobelx^2 + sobely^2)
    sobel_combined = np.sqrt(sobelx**2 + sobely**2)

    # 4. Normalize the combined image for better visualization
    sobel_combined = cv2.normalize(sobel_combined, None, 0, 255, cv2.NORM_MINMAX, cv2.CV_8U)

    # 5. Display the original and edge-detected images
    print("Original Image")
    cv2_imshow(img)

    print("Sobel Edges")
    cv2_imshow(sobel_combined)

    # Alternatively, you can plot them using matplotlib
    # Note: matplotlib expects RGB, but we are using grayscale
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.title('Original Image')
    plt.imshow(img, cmap='gray')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.title('Sobel Edges')
    plt.imshow(sobel_combined, cmap='gray')
    plt.axis('off')

    plt.show()
```

```python
# Import necessary libraries
import cv2
from google.colab.patches import cv2_imshow
import numpy as np


img = cv2.imread('/content/drive/MyDrive/Applied ML/img1.jpg', 1)
# Check if the image was loaded correctly
if img is None:
    print("Error: Could not load image.")
else:
    # Display the original image
    print("Original Image:")
    cv2_imshow(img)

    # 1. Convert the image to grayscale
    # The Laplacian operator works on single-channel images.
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 2. Apply the Laplacian filter
    laplacian = cv2.Laplacian(gray, cv2.CV_64F, ksize=3)

    # 3. Convert the output back to an 8-bit unsigned integer

    laplacian_abs = cv2.convertScaleAbs(laplacian)

    # Display the result
    print("\nLaplacian Edge Detection Result:")
    cv2_imshow(laplacian_abs)
```

```python
# Import necessary libraries
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread('/content/drive/MyDrive/Applied ML/img1.jpg', 1)

# Check if the image loaded correctly
if img is None:
    print("Error: Could not load image.")
else:
    # Display the original image
    print("Original Image:")
    cv2_imshow(img)

    # 1. Convert the image to grayscale
    # Corner detection works on a single-channel image.
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 2. Convert to float32
    # The Harris corner detector expects a float32 image.
    gray = np.float32(gray)

    # 3. Apply the Harris corner detector
    dst = cv2.cornerHarris(gray, 2, 3, 0.04)

    # 4. Dilate the corner response image

    dst = cv2.dilate(dst, None)

    # 5. Threshold the response to get corner points

    img[dst > 0.01 * dst.max()] = [0, 0, 255] # Mark corners with a red color

    # Display the final image with detected corners
    print("\nImage with Detected Corners:")
    cv2_imshow(img)
```

```python
# Import necessary libraries
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
import requests
from io import BytesIO

# Download the image using requests and convert it to a NumPy array
try:
    img = cv2.imread('/content/drive/MyDrive/Applied ML/img1.jpg',1)

    # 1. Convert the image to grayscale
    # SIFT works best on single-channel images.
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 2. Create a SIFT detector object
    # cv2.SIFT_create() is the function for SIFT in OpenCV
    sift = cv2.SIFT_create()

    # 3. Detect keypoints
    # The detect() function finds the keypoints in the image.
    keypoints = sift.detect(gray, None)

    # 4. Draw the keypoints on the original image
    # cv2.drawKeypoints(image, keypoints, output_image, flags)
    # The flag cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS ensures that the
    # circle and orientation are drawn for each keypoint.
    output_image = cv2.drawKeypoints(img, keypoints, None, flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

    # 5. Display the final image
    print("Original Image:")
    cv2_imshow(img)
    print("\nImage with Detected SIFT Keypoints:")
    cv2_imshow(output_image)

except Exception as e:
    print(f"An error occurred: {e}")
```
