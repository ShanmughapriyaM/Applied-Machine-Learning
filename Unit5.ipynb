```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.decomposition import PCA, KernelPCA

# Step 1: Create nonlinear dataset
X, y = make_moons(n_samples=200, noise=0.05, random_state=42)

# Step 2: Apply Standard PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Step 3: Apply Kernel PCA (RBF kernel)
kpca = KernelPCA(n_components=2, kernel="rbf", gamma=15)
X_kpca = kpca.fit_transform(X)

# Step 4: Make a comparison table
df = pd.DataFrame({
    "PCA_1": X_pca[:,0],
    "PCA_2": X_pca[:,1],
    "KPCA_1": X_kpca[:,0],
    "KPCA_2": X_kpca[:,1],
    "Class": y
})

print(df.head(10))  # Show first 10 rows

# Step 5: Plot both results
plt.figure(figsize=(12,5))

# Plot PCA
plt.subplot(1,2,1)
plt.scatter(X_pca[:,0], X_pca[:,1], c=y, cmap="coolwarm", edgecolor="k")
plt.title("Linear PCA (2D)")
plt.xlabel("PC1"); plt.ylabel("PC2")

# Plot Kernel PCA
plt.subplot(1,2,2)
plt.scatter(X_kpca[:,0], X_kpca[:,1], c=y, cmap="coolwarm", edgecolor="k")
plt.title("Kernel PCA (RBF, 2D)")
plt.xlabel("KPC1"); plt.ylabel("KPC2")

plt.show()
```
